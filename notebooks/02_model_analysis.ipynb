# Model Evaluation Summary

A total of six machine learning algorithms were evaluated for Nepali news classification.

- Naive Bayes
- Logistic Regression
- Linear SVM
- Random Forest
- Decision Tree
- K-Nearest Neighbors (KNN)

Among these, **Linear SVM** achieved the highest overall performance.

## Best Model

- Model: Linear SVM
- Accuracy: 80.31%
- Precision: 78.07%
- Recall: 80.31%
- F1 Score: 78.58%

The confusion matrix shows that most categories are correctly classified, with some misclassification occurring between semantically similar categories such as General News and Politics & Government.

Overall, the Linear SVM model provides the best balance between prediction accuracy and computational efficiency for this Nepali news classification task.
